## Implement Tools Screen Logic

### Subtask:
Develop Python code for the Driving Timer functionality (Start, Pause, Reset) and a basic logbook system to store and retrieve notes.


**Reasoning**:
I will create a Python code block to define the `ToolsScreen` class with driving timer and logbook functionalities, as well as demonstrate their usage as requested in the instructions.



````python
import time
import datetime

class ToolsScreen:
    def __init__(self):
        self.timer_start_time = None
        self.paused_duration = 0  # Total time accumulated when paused
        self.is_running = False
        self.logbook_entries = []

    def start_timer(self):
        """Starts or resumes the driving timer."""
        if not self.is_running:
            self.timer_start_time = time.time()
            self.is_running = True
            print("Timer started.")
        else:
            print("Timer is already running.")

    def pause_timer(self):
        """Pauses the driving timer and accumulates the elapsed time."""
        if self.is_running:
            current_elapsed_time = time.time() - self.timer_start_time
            self.paused_duration += current_elapsed_time
            self.is_running = False
            print("Timer paused. Current total driving time accumulated:" \
                  f" {str(datetime.timedelta(seconds=int(self.paused_duration)))}.")
        else:
            print("Timer is not running to be paused.")

    def reset_timer(self):
        """Resets the driving timer and clears accumulated time."""
        self.timer_start_time = None
        self.paused_duration = 0
        self.is_running = False
        print("Timer reset.")

    def get_current_time(self):
        """Returns the current elapsed driving time, accounting for pauses."""
        if self.is_running:
            current_total_time = self.paused_duration + (time.time() - self.timer_start_time)
        else:
            current_total_time = self.paused_duration

        # Format as H:MM:SS
        formatted_time = str(datetime.timedelta(seconds=int(current_total_time)))
        print(f"Current driving time: {formatted_time}")
        return formatted_time

    def add_log_entry(self, note):
        """Appends a timestamped note to the logbook."""
        timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        entry = f"[{timestamp}] {note}"
        self.logbook_entries.append(entry)
        print(f"Log entry added: {entry}")

    def view_log_entries(self):
        """Returns all stored log entries."""
        if not self.logbook_entries:
            print("No log entries yet.")
            return []
        else:
            print("--- Logbook Entries ---")
            for entry in self.logbook_entries:
                print(entry)
            print("-----------------------")
            return self.logbook_entries

# Example Usage:
tools_screen = ToolsScreen()

print("\n--- Demonstrating Driving Timer ---")
# 1. Start timer
tools_screen.start_timer()
time.sleep(2) # Simulate 2 seconds of driving
tools_screen.get_current_time()

# 2. Pause timer
tools_screen.pause_timer()
time.sleep(1) # Simulate 1 second while paused
tools_screen.get_current_time()

# 3. Resume timer
tools_screen.start_timer()
time.sleep(3) # Simulate another 3 seconds of driving
tools_screen.get_current_time()

# 4. Reset timer
tools_screen.reset_timer()
tools_screen.get_current_time()

print("\n--- Demonstrating Logbook Functionality ---")
# 1. Add log entries
tools_screen.add_log_entry("Started shift in Los Angeles.")
time.sleep(0.5)
tools_screen.add_log_entry("Took a 30 minute break.")
time.sleep(0.5)
tools_screen.add_log_entry("Arrived at destination city.")

# 2. View log entries
tools_screen.view_log_entries()
````

## Build Profile Management Logic

### Subtask:
Create Python functions to handle saving and loading driver profile information (Name, Company, Home Terminal) locally, and to generate Google Maps sharing links for current coordinates.


**Reasoning**:
I will create a Python class `ProfileManagement` with methods to save, load, and display driver information, and generate Google Maps sharing links, then demonstrate its usage with example data, as outlined in the instructions.



````python
import json
import os

class ProfileManagement:
    def __init__(self, filename="driver_profile.json"):
        self.driver_info = {
            "Name": None,
            "Company": None,
            "Home Terminal": None
        }
        self.filename = filename
        self._load_profile_from_file() # Attempt to load existing profile on instantiation

    def _load_profile_from_file(self):
        if os.path.exists(self.filename):
            try:
                with open(self.filename, 'r') as f:
                    loaded_info = json.load(f)
                    self.driver_info.update(loaded_info)
                print(f"Profile loaded from {self.filename}")
            except json.JSONDecodeError:
                print(f"Error decoding JSON from {self.filename}. Starting with empty profile.")
        else:
            print(f"No existing profile found at {self.filename}. Starting with empty profile.")

    def save_driver_info(self, name, company, home_terminal):
        """Saves driver information locally."""
        self.driver_info["Name"] = name
        self.driver_info["Company"] = company
        self.driver_info["Home Terminal"] = home_terminal

        try:
            with open(self.filename, 'w') as f:
                json.dump(self.driver_info, f, indent=4)
            print("Driver information saved successfully.")
        except IOError as e:
            print(f"Error saving profile to file: {e}")

    def load_driver_info(self):
        """Loads and returns the saved driver information."""
        print("--- Current Driver Profile ---")
        if all(value is not None for value in self.driver_info.values()):
            for key, value in self.driver_info.items():
                print(f"{key}: {value}")
        else:
            print("Profile information incomplete or not set.")
            for key, value in self.driver_info.items():
                print(f"{key}: {value if value is not None else 'Not set'}")
        print("------------------------------")
        return self.driver_info

    @staticmethod
    def generate_location_sharing_url(latitude, longitude):
        """Generates a Google Maps location sharing URL."""
        url = f"https://www.google.com/maps/place/{latitude},{longitude}"
        print(f"Generated Location Sharing URL: {url}")
        return url

# Example Usage:

# 1. Instantiate the ProfileManagement class
profile_manager = ProfileManagement()

print("\n--- Demonstrating Profile Management ---")

# 2. Save some example driver information
profile_manager.save_driver_info("John Doe", "Trucking Co. A", "Dallas, TX")

# 3. Load and print the saved driver information
loaded_info = profile_manager.load_driver_info()

# 4. Generate and print a location sharing link using example GPS coordinates
example_latitude = 34.052235
example_longitude = -118.243683
location_link = profile_manager.generate_location_sharing_url(example_latitude, example_longitude)

# Demonstrate loading again to show persistence (if run in separate session)
print("\n--- Demonstrating Profile Load after potential re-instantiation ---")
profile_manager_reloaded = ProfileManagement()
profile_manager_reloaded.load_driver_info()
````

````python
import time

class HomeDashboard:
    def __init__(self):
        self.current_latitude = None
        self.current_longitude = None
        self.origin = None
        self.destination = None
        # New attributes for weather, traffic, and services
        self.current_weather_data = None
        self.traffic_alert_data = None
        self.nearby_services_data = None

    def get_gps_location(self):
        """Simulates fetching the current GPS coordinates."""
        self.current_latitude = 34.052235  # Example: Los Angeles latitude
        self.current_longitude = -118.243683 # Example: Los Angeles longitude
        print(f"Simulated GPS Location: ({self.current_latitude}, {self.current_longitude})")
        return (self.current_latitude, self.current_longitude)

    def set_origin_destination(self, origin, destination):
        """Accepts and stores origin and destination strings."""
        self.origin = origin
        self.destination = destination
        print(f"Origin set to: {self.origin}, Destination set to: {self.destination}")

    def generate_google_maps_url(self, search_term):
        """Generates a Google Maps search URL for quick buttons."""
        if self.current_latitude and self.current_longitude:
            location_param = f"{self.current_latitude},{self.current_longitude}"
            base_url = "https://www.google.com/maps/search/"
            formatted_search_term = search_term.replace(' ', '+')
            url = f"{base_url}{formatted_search_term}+near+{location_param}"
            print(f"Generated Google Maps URL for '{search_term}': {url}")
            return url
        else:
            print("GPS location not set. Cannot generate location-aware URL.")
            return f"https://www.google.com/maps/search/{search_term.replace(' ', '+')}"

    def fetch_weather_data(self, latitude, longitude):
        """Simulates fetching current weather conditions for given coordinates."""
        # In a real app, this would use a weather API
        self.current_weather_data = {
            "location": f"{latitude}, {longitude}",
            "conditions": "Partly Cloudy",
            "temperature_c": 22,
            "temperature_f": 72,
            "forecast": "Tomorrow: Sunny, High 75F",
            "alerts": []
        }
        print(f"Simulated weather data fetched for {latitude}, {longitude}.")

    def display_weather(self):
        """Prints the simulated weather data in a user-friendly format."""
        if self.current_weather_data:
            print("\n--- Current Weather ---")
            print(f"Location: {self.current_weather_data['location']}")
            print(f"Conditions: {self.current_weather_data['conditions']}")
            print(f"Temperature: {self.current_weather_data['temperature_f']}F ({self.current_weather_data['temperature_c']}C)")
            print(f"Forecast: {self.current_weather_data['forecast']}")
            if self.current_weather_data['alerts']:
                print(f"Alerts: {', '.join(self.current_weather_data['alerts'])}")
            print("-----------------------")
        else:
            print("No weather data available. Please fetch first.")

    def fetch_traffic_alerts(self, origin=None, destination=None, radius=None):
        """Simulates fetching real-time traffic updates."""
        # In a real app, this would use a traffic API
        if origin and destination:
            self.traffic_alert_data = [
                f"Heavy congestion on route from {origin} to {destination}. Expect 30 min delay.",
                "Accident reported on I-5 South near Exit 10. Left lane blocked."
            ]
            print(f"Simulated traffic alerts fetched for route from {origin} to {destination}.")
        elif radius and self.current_latitude and self.current_longitude:
             self.traffic_alert_data = [
                f"Minor delays within {radius} miles of {self.current_latitude}, {self.current_longitude}.",
                "Construction on Main St. causing detours."
            ]
             print(f"Simulated traffic alerts fetched within {radius} miles of current location.")
        else:
            self.traffic_alert_data = []
            print("No specific route or radius provided for traffic alerts.")

    def display_traffic_alerts(self):
        """Prints the simulated traffic alerts."""
        if self.traffic_alert_data:
            print("\n--- Traffic Alerts ---")
            for alert in self.traffic_alert_data:
                print(f"- {alert}")
            print("----------------------")
        else:
            print("No traffic alerts available.")

    def find_nearby_services(self, latitude, longitude, service_type=None, radius=50):
        """Simulates finding truck-specific services near given coordinates."""
        # In a real app, this would use a places API (e.g., Google Places)
        self.nearby_services_data = []
        if service_type == "repair shops":
            self.nearby_services_data = [
                {"name": "Truck Repair Express", "address": "123 Industrial Rd", "contact": "555-1234", "location": "34.05, -118.2"},
                {"name": "Heavy Duty Service Center", "address": "456 Commerce St", "contact": "555-5678", "location": "34.07, -118.25"}
            ]
        elif service_type == "wash stations":
             self.nearby_services_data = [
                {"name": "Big Rig Wash", "address": "789 Trucker Ave", "contact": "555-9012", "location": "34.06, -118.21"}
            ]
        elif service_type is None:
             self.nearby_services_data = [
                {"name": "Truck Repair Express", "address": "123 Industrial Rd", "contact": "555-1234", "location": "34.05, -118.2"},
                {"name": "Big Rig Wash", "address": "789 Trucker Ave", "contact": "555-9012", "location": "34.06, -118.21"}
            ]

        print(f"Simulated nearby '{service_type if service_type else 'all'}' services found near {latitude}, {longitude} within {radius} miles.")

    def display_nearby_services(self):
        """Prints the simulated nearby services."""
        if self.nearby_services_data:
            print("\n--- Nearby Services ---")
            for service in self.nearby_services_data:
                print(f"- {service['name']} at {service['address']} (Contact: {service['contact']})")
            print("-----------------------")
        else:
            print("No nearby services available.")

    def generate_service_direction_url(self, service_location):
        """Generates a Google Maps URL for directions to a service location."""
        # Assuming service_location is a string like "latitude,longitude" or an address
        encoded_location = service_location.replace(' ', '+')
        url = f"https://www.google.com/maps/dir/?api=1&destination={encoded_location}"
        print(f"Generated direction URL for '{service_location}': {url}")
        return url

# Example Usage with new features:
home_dashboard = HomeDashboard()

print("\n--- Home Dashboard - New Features Demo ---")

# 1. Simulate GPS location
gps_coords = home_dashboard.get_gps_location()
current_lat, current_lon = gps_coords

# 2. Fetch and display weather data
home_dashboard.fetch_weather_data(current_lat, current_lon)
home_dashboard.display_weather()

# 3. Fetch and display traffic alerts
home_dashboard.set_origin_destination("Los Angeles, CA", "San Francisco, CA")
home_dashboard.fetch_traffic_alerts(origin=home_dashboard.origin, destination=home_dashboard.destination)
home_dashboard.display_traffic_alerts()

# 4. Find and display nearby services
home_dashboard.find_nearby_services(current_lat, current_lon, service_type="repair shops")
home_dashboard.display_nearby_services()

# 5. Generate service direction URL
if home_dashboard.nearby_services_data:
    first_service_location = home_dashboard.nearby_services_data[0]['location'] # Using simulated location
    home_dashboard.generate_service_direction_url(first_service_location)

print("\n--- Old Home Dashboard Functionality Still Works ---")
# 6. Set Origin and Destination (from previous step, re-run for context)
home_dashboard.set_origin_destination("Los Angeles, CA", "San Francisco, CA")

# 7. Generate Google Maps URLs (from previous step, re-run for context)
truck_stops_url = home_dashboard.generate_google_maps_url("Truck Stops")
fuel_stations_url = home_dashboard.generate_google_maps_url("Fuel Stations")
````

````python
import time
import datetime
import asyncio

class FlutterDrivingTimer:
    def __init__(self):
        self._timer_start_time = None
        self._paused_duration = 0  # Total time accumulated when paused
        self._is_running = False
        print("FlutterDrivingTimer initialized.")

    async def start_timer(self):
        """Simulates starting or resuming the driving timer in Flutter."""
        if not self._is_running:
            self._timer_start_time = time.time()
            self._is_running = True
            print("Timer started (Flutter sim).")
        else:
            print("Timer is already running (Flutter sim).")

    async def pause_timer(self):
        """Simulates pausing the driving timer in Flutter."""
        if self._is_running:
            current_elapsed_time = time.time() - self._timer_start_time
            self._paused_duration += current_elapsed_time
            self._is_running = False
            print("Timer paused (Flutter sim). Current total driving time accumulated:" \
                  f" {str(datetime.timedelta(seconds=int(self._paused_duration)))}.")
        else:
            print("Timer is not running to be paused (Flutter sim).")

    async def reset_timer(self):
        """Simulates resetting the driving timer in Flutter."""
        self._timer_start_time = None
        self._paused_duration = 0
        self._is_running = False
        print("Timer reset (Flutter sim).")

    async def get_current_time(self):
        """Simulates returning the current elapsed driving time, accounting for pauses."""
        if self._is_running:
            current_total_time = self._paused_duration + (time.time() - self._timer_start_time)
        else:
            current_total_time = self._paused_duration

        # Format as H:MM:SS
        formatted_time = str(datetime.timedelta(seconds=int(current_total_time)))
        print(f"Current driving time (Flutter sim): {formatted_time}")
        return formatted_time


# Example Usage:
async def demo_flutter_driving_timer():
    timer = FlutterDrivingTimer()

    print("\n--- Demonstrating Flutter-like Driving Timer ---")
    # 1. Start timer
    await timer.start_timer()
    await asyncio.sleep(2) # Simulate 2 seconds of driving
    await timer.get_current_time()

    # 2. Pause timer
    await timer.pause_timer()
    await asyncio.sleep(1) # Simulate 1 second while paused
    await timer.get_current_time()

    # 3. Resume timer
    await timer.start_timer()
    await asyncio.sleep(3) # Simulate another 3 seconds of driving
    await timer.get_current_time()

    # 4. Reset timer
    await timer.reset_timer()
    await timer.get_current_time()

# Check if an asyncio event loop is already running
try:
    loop = asyncio.get_running_loop()
except RuntimeError:
    loop = None

if loop and loop.is_running():
    # If a loop is running, run the coroutine until complete
    task = loop.create_task(demo_flutter_driving_timer())
    print("Async task for Flutter-like driving timer scheduled.")
else:
    # If no loop is running, start a new one
    asyncio.run(demo_flutter_driving_timer())
````

Develop the user interface for all four screens (Splash, Home Dashboard, Tools, and Profile) using Flutter, ensuring smooth transitions, responsive design, and adherence to the styling guidelines. Implement the navigation flow between screens.

#### Instructions
1. Set up a new Flutter project to begin the frontend development, configuring `pubspec.yaml` for necessary dependencies (e.g., state management, geolocator, http, url_launcher) and asset management.
2. Develop the UI for the Splash Screen (`splash_screen.dart`), including the app logo (`Image.asset`) and a loading indicator (`CircularProgressIndicator` or custom animation). Implement a 3-second delay using `Future.delayed` followed by an automatic transition to the Home Dashboard using `Navigator.of(context).pushReplacement`.
3. Implement the Home Dashboard UI (`home_screen.dart`) with a `Scaffold`, `AppBar`, `Column` or `ListView` layout. Include `Text` widgets for GPS status, `TextField` widgets for origin/destination, and `Row` or `Wrap` of `ElevatedButton` widgets for quick searches (Truck Stops, Fuel, Restaurants, Scales, Diesel). Incorporate `Card` or custom widgets for weather display (conditions, temperature, forecast), a `Container` or `SnackBar` for traffic alerts, and an `ElevatedButton` or `ListView` for nearby truck services. Ensure all interactive elements, like buttons, are rounded and use the `#FE8C10` accent color.
4. Build the UI for the Tools Screen (`tools_screen.dart`) with a `Scaffold`. This should include `Text` widgets for the driving timer display, `ElevatedButton` widgets for 'Start', 'Pause', and 'Reset'. Add a `TextField` for new logbook notes and a `ListView.builder` to display past entries. Include distinct `ElevatedButton` widgets for 'DOT Information' and 'CDL School' web searches.
5. Create the UI for Profile Management (`profile_screen.dart`) with a `Scaffold`. This should contain a `Form` widget with `TextFormField` widgets for 'Name', 'Company', and 'Home Terminal', an `ElevatedButton` for 'Save', and another `ElevatedButton` for 'Generate Location Sharing Link'.
6. For all screens, ensure the design strictly adheres to the 'Modern Dark Theme' for `backgroundColor` and uses clean, modern sans-serif fonts for all text elements.
7. Implement the navigation flow and transitions between the Splash Screen, Home Dashboard, Tools Screen, and Profile Management. Use Flutter's `Navigator` for transitions, setting the Splash Screen as the initial route and considering a `BottomNavigationBar` on the main screens for seamless switching.

I will create a Python class `FlutterDrivingTimer` to conceptually simulate the driving timer functionality in Flutter, demonstrating how start, pause, reset, and current time tracking would work.



````python
import time
import datetime
import asyncio

class FlutterDrivingTimer:
    def __init__(self):
        self._timer_start_time = None
        self._paused_duration = 0  # Total time accumulated when paused
        self._is_running = False
        print("FlutterDrivingTimer initialized.")

    async def start_timer(self):
        """Simulates starting or resuming the driving timer in Flutter."""
        if not self._is_running:
            self._timer_start_time = time.time()
            self._is_running = True
            print("Timer started (Flutter sim).")
        else:
            print("Timer is already running (Flutter sim).")

    async def pause_timer(self):
        """Simulates pausing the driving timer in Flutter."""
        if self._is_running:
            current_elapsed_time = time.time() - self._timer_start_time
            self._paused_duration += current_elapsed_time
            self._is_running = False
            print("Timer paused (Flutter sim). Current total driving time accumulated:" \
                  f" {str(datetime.timedelta(seconds=int(self._paused_duration)))}.")
        else:
            print("Timer is not running to be paused (Flutter sim).")

    async def reset_timer(self):
        """Simulates resetting the driving timer in Flutter."""
        self._timer_start_time = None
        self._paused_duration = 0
        self._is_running = False
        print("Timer reset (Flutter sim).")

    async def get_current_time(self):
        """Simulates returning the current elapsed driving time, accounting for pauses."""
        if self._is_running:
            current_total_time = self._paused_duration + (time.time() - self._timer_start_time)
        else:
            current_total_time = self._paused_duration

        # Format as H:MM:SS
        formatted_time = str(datetime.timedelta(seconds=int(current_total_time)))
        print(f"Current driving time (Flutter sim): {formatted_time}")
        return formatted_time


# Example Usage:
async def demo_flutter_driving_timer():
    timer = FlutterDrivingTimer()

    print("\n--- Demonstrating Flutter-like Driving Timer ---")
    # 1. Start timer
    await timer.start_timer()
    await asyncio.sleep(2) # Simulate 2 seconds of driving
    await timer.get_current_time()

    # 2. Pause timer
    await timer.pause_timer()
    await asyncio.sleep(1) # Simulate 1 second while paused
    await timer.get_current_time()

    # 3. Resume timer
    await timer.start_timer()
    await asyncio.sleep(3) # Simulate another 3 seconds of driving
    await timer.get_current_time()

    # 4. Reset timer
    await timer.reset_timer()
    await timer.get_current_time()

# Check if an asyncio event loop is already running
try:
    loop = asyncio.get_running_loop()
except RuntimeError:
    loop = None

if loop and loop.is_running():
    # If a loop is running, run the coroutine until complete
    task = loop.create_task(demo_flutter_driving_timer())
    print("Async task for Flutter-like driving timer scheduled.")
else:
    # If no loop is running, start a new one
    asyncio.run(demo_flutter_driving_timer())
````
